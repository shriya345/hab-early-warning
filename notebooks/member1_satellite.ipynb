```python
from datetime import datetime

# Target approximately every 10 days:
# 1st, 11th and 21st of every month

target_dates = []

for month in range(1, 13):
    for day in [1, 11, 21]:
        target_dates.append(
            datetime(2024, month, day)
        )

print("Total target dates:", len(target_dates))

for d in target_dates:
    print(d.strftime("%Y-%m-%d"))
```

```python
selected = []

available_dates = sorted(
    datetime.strptime(d, "%Y-%m-%d")
    for d in tif_by_date.keys()
    if d.startswith("2024-")
)

MAX_GAP_DAYS = 5

used_dates = set()

for target in target_dates:

    candidates = [
        d for d in available_dates
        if abs((d - target).days) <= MAX_GAP_DAYS
        and d.strftime("%Y-%m-%d") not in used_dates
    ]

    if candidates:
        closest = min(
            candidates,
            key=lambda d: abs((d - target).days)
        )

        actual_date = closest.strftime("%Y-%m-%d")

        selected.append({
            "target_date": target.strftime("%Y-%m-%d"),
            "actual_date": actual_date,
            "gap_days": abs((closest - target).days),
            "file": tif_by_date[actual_date][0]
        })

        used_dates.add(actual_date)

print("Selected images:", len(selected))
```

```python
available_dates = sorted(
    pd.to_datetime(list(tif_by_date.keys()))
)

results = []

for target in target_dates:

    closest = min(
        available_dates,
        key=lambda x: abs(x - target)
    )

    gap = abs((closest - target).days)

    results.append({
        "target_date": target.strftime("%Y-%m-%d"),
        "actual_date": closest.strftime("%Y-%m-%d"),
        "gap_days": gap,
        "tiff_file": tif_by_date[closest.strftime("%Y-%m-%d")][0]
    })

validation_df = pd.DataFrame(results)

print(validation_df.to_string(index=False))
```

```python
import pandas as pd

# All available TIFF dates
available_dates = sorted(pd.to_datetime(list(tif_by_date.keys())))

# Target dates: every 10 days through 2024
target_dates = pd.date_range(
    start="2024-01-01",
    end="2024-12-31",
    freq="10D"
)

results = []
used_dates = set()

for target in target_dates:

    # Find unused TIFF dates
    candidates = [
        d for d in available_dates
        if d.date() not in used_dates
    ]

    if not candidates:
        results.append({
            "target_date": target.date(),
            "actual_date": None,
            "gap_days": None,
            "status": "NO TIFF AVAILABLE"
        })
        continue

    # Find nearest unused TIFF
    nearest = min(
        candidates,
        key=lambda d: abs((d - target).days)
    )

    gap = abs((nearest - target).days)

    # Accept only if within 5 days of target
    if gap <= 5:
        used_dates.add(nearest.date())

        results.append({
            "target_date": target.date(),
            "actual_date": nearest.date(),
            "gap_days": gap,
            "status": "VALID"
        })

    else:
        results.append({
            "target_date": target.date(),
            "actual_date": nearest.date(),
            "gap_days": gap,
            "status": "NO TIFF WITHIN ±5 DAYS"
        })

validation_df = pd.DataFrame(results)

print(validation_df.to_string(index=False))
```

```python
valid_df = validation_df[
    validation_df["status"] == "VALID"
].copy()

valid_df["actual_date"] = pd.to_datetime(valid_df["actual_date"])

valid_df["gap_from_previous"] = (
    valid_df["actual_date"].diff().dt.days
)

print(valid_df.to_string(index=False))
```

```python
spacing_problems = valid_df[
    valid_df["gap_from_previous"].notna() &
    (valid_df["gap_from_previous"] < 10)
]

print("Images less than 10 days apart:")
print(spacing_problems.to_string(index=False))
```

```python
available_dates = sorted(
    pd.to_datetime(list(tif_by_date.keys()))
)

used_dates = set()
results = []

for target in target_dates:

    # Find TIFFs within ±5 days of target
    candidates = [
        d for d in available_dates
        if abs((d - target).days) <= 5
        and d.date() not in used_dates
    ]

    if candidates:

        # Choose the closest available TIFF
        selected = min(
            candidates,
            key=lambda d: abs((d - target).days)
        )

        used_dates.add(selected.date())

        results.append({
            "target_date": target.date(),
            "actual_date": selected.date(),
            "gap_days": abs((selected - target).days),
            "status": "VALID",
            "tiff_path": tif_by_date[selected.strftime("%Y-%m-%d")]
        })

    else:

        results.append({
            "target_date": target.date(),
            "actual_date": None,
            "gap_days": None,
            "status": "MISSING",
            "tiff_path": None
        })

validation_df = pd.DataFrame(results)

print(validation_df.to_string(index=False))
```
