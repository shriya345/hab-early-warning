```python
import xarray as xr
import pandas as pd
import glob
import numpy as np
import os

# Find all downloaded ERA5 files
files = sorted(glob.glob("era5_vembanad*.nc"))

print("Files found:")
for f in files:
    print(f)

# Open and combine all monthly files
ds = xr.open_mfdataset(
    files,
    combine="by_coords",
    engine="netcdf4"
)

print("\nDataset loaded successfully!")
print(ds)

# Take spatial average for the Vembanad Lake area
ds_mean = ds.mean(dim=["latitude", "longitude"])

# Calculate hourly wind speed
wind_speed = np.sqrt(
    ds_mean["u10"]**2 +
    ds_mean["v10"]**2
)

# Create a new dataset
processed = xr.Dataset({
    "rainfall": ds_mean["tp"] * 1000,   # Convert metres to mm
    "wind_speed": wind_speed
})

# Convert hourly data to daily data
daily = xr.Dataset({
    "rainfall": processed["rainfall"].resample(valid_time="1D").sum(),
    "wind_speed": processed["wind_speed"].resample(valid_time="1D").mean()
})

# Convert to Pandas DataFrame
df = daily.to_dataframe().reset_index()

# Keep only required columns
df = df[["valid_time", "rainfall", "wind_speed"]]

# Rename date column
df = df.rename(columns={"valid_time": "date"})

# Save CSV
df.to_csv("vembanad_rainfall_wind_daily.csv", index=False)

print("\nSUCCESS!")
print("CSV created: vembanad_rainfall_wind_daily.csv")
print("\nFirst 5 rows:")
print(df.head())
```

```python
import xarray as xr
import pandas as pd
import glob
import numpy as np
import os

# Find all downloaded ERA5 files
files = sorted(glob.glob("era5_vembanad*.nc"))

print("Number of files found:", len(files))

for f in files:
    print(f)

# Open files one by one
datasets = []
valid_files = [] # To keep track of successfully opened files

for file in files:
    try:
        ds = xr.open_dataset(file, engine="netcdf4")
        datasets.append(ds)
        valid_files.append(file)
    except OSError as e:
        print(f"WARNING: Could not open {file} due to OSError: {e}. This file will be skipped.")
        print(f"It is recommended to delete this file and re-download it. File path: {file}")

if not datasets:
    raise ValueError("No valid NetCDF files could be opened. Please check your downloaded files.")

print(f"\nSuccessfully opened {len(datasets)} files out of {len(files)} total files.")
print("Successfully opened files:")
for f in valid_files:
    print(f)

# Combine all monthly datasets
ds = xr.combine_by_coords(datasets)

print("\nAll datasets combined successfully!")
print(ds)

# Find the time column automatically
if "valid_time" in ds.dims:
    time_dim = "valid_time"
else:
    time_dim = "time"

# Find spatial dimensions
spatial_dims = []

if "latitude" in ds.dims:
    spatial_dims.append("latitude")

if "longitude" in ds.dims:
    spatial_dims.append("longitude")

# Take average of the Vembanad area
ds_mean = ds.mean(dim=spatial_dims)

# Calculate wind speed
wind_speed = np.sqrt(
    ds_mean["u10"] ** 2 +
    ds_mean["v10"] ** 2
)

# Create processed dataset
processed = xr.Dataset({
    "rainfall": ds_mean["tp"] * 1000,
    "wind_speed": wind_speed
})

# Convert hourly data to daily data
daily_rainfall = processed["rainfall"].resample(
    {time_dim: "1D"}
).sum()

daily_wind = processed["wind_speed"].resample(
    {time_dim: "1D"}
).mean()

# Create final daily dataset
daily = xr.Dataset({
    "rainfall": daily_rainfall,
    "wind_speed": daily_wind
})

# Convert to DataFrame
df = daily.to_dataframe().reset_index()

# Rename time column to date
df = df.rename(columns={time_dim: "date"})

# Keep required columns
df = df[["date", "rainfall", "wind_speed"]]

# Save CSV
df.to_csv(
    "vembanad_rainfall_wind_daily.csv",
    index=False
)

print("\nSUCCESS! 🎉")
print("CSV created: vembanad_rainfall_wind_daily.csv")

print("\nFirst 5 rows:")
print(df.head())

print("\nTotal rows:", len(df))
```

```python
import xarray as xr
import pandas as pd
import numpy as np
import glob

# =========================================================
# 1. FIND ALL 12 MONTHLY FILES
# =========================================================

instant_files = sorted(
    glob.glob("era5_monthly/*/data_stream-oper_stepType-instant.nc")
)

accum_files = sorted(
    glob.glob("era5_monthly/*/data_stream-oper_stepType-accum.nc")
)

print("Instant files found:", len(instant_files))
print("Accumulation files found:", len(accum_files))


# =========================================================
# 2. PROCESS WIND DATA (u10 and v10)
# =========================================================

wind_frames = []

for file in instant_files:
    print("Processing wind:", file)

    ds = xr.open_dataset(file, engine="netcdf4")

    # Spatial average
    ds_mean = ds[["u10", "v10"]].mean(
        dim=["latitude", "longitude"]
    )

    # Wind speed calculation
    wind_speed = np.sqrt(
        ds_mean["u10"] ** 2 +
        ds_mean["v10"] ** 2
    )

    # Convert to DataFrame
    wind_df = wind_speed.to_dataframe(
        name="wind_speed"
    ).reset_index()

    wind_frames.append(wind_df)

    ds.close()

# Combine all months
wind_df = pd.concat(
    wind_frames,
    ignore_index=True
)

# Remove duplicate dates if any
wind_df = wind_df.drop_duplicates(
    subset=["valid_time"]
)

# Sort by time
wind_df = wind_df.sort_values("valid_time")

# Daily average wind speed
daily_wind = (
    wind_df
    .set_index("valid_time")
    .resample("D")["wind_speed"]
    .mean()
    .reset_index()
)


# =========================================================
# 3. PROCESS RAINFALL DATA (tp)
# =========================================================

rain_frames = []

for file in accum_files:
    print("Processing rainfall:", file)

    ds = xr.open_dataset(file, engine="netcdf4")

    # Spatial average
    tp_mean = ds["tp"].mean(
        dim=["latitude", "longitude"]
    )

    # Convert metres to millimetres
    tp_mm = tp_mean * 1000

    # Convert to DataFrame
    rain_df = tp_mm.to_dataframe(
        name="rainfall"
    ).reset_index()

    rain_frames.append(rain_df)

    ds.close()

# Combine all months
rain_df = pd.concat(
    rain_frames,
    ignore_index=True
)

# Remove duplicate dates if any
rain_df = rain_df.drop_duplicates(
    subset=["valid_time"]
)

# Sort by time
rain_df = rain_df.sort_values("valid_time")

# Daily total rainfall
daily_rain = (
    rain_df
    .set_index("valid_time")
    .resample("D")["rainfall"]
    .sum()
    .reset_index()
)


# =========================================================
# 4. COMBINE RAINFALL + WIND
# =========================================================

final_df = pd.merge(
    daily_rain,
    daily_wind,
    on="valid_time",
    how="inner"
)

# Rename date column
final_df = final_df.rename(
    columns={"valid_time": "date"}
)

# Keep only required columns
final_df = final_df[
    ["date", "rainfall", "wind_speed"]
]

# Round values
final_df["rainfall"] = final_df["rainfall"].round(3)
final_df["wind_speed"] = final_df["wind_speed"].round(3)


# =========================================================
# 5. SAVE FINAL CSV
# =========================================================

output_file = "vembanad_2024_rainfall_wind.csv"

final_df.to_csv(
    output_file,
    index=False
)

print("\n" + "=" * 50)
print("SUCCESS! FINAL CSV CREATED 🎉")
print("=" * 50)

print("File name:", output_file)
print("Total days:", len(final_df))

print("\nFirst 10 rows:")
print(final_df.head(10))

print("\nLast 5 rows:")
print(final_df.tail())

print("\nColumns:")
print(final_df.columns.tolist())
```
